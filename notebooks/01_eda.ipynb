4. Проверка дубликатов

```python
import pandas as pd

df = pd.read_csv("../data/creditcard.csv")

print("Первые пять строк:")
print(df.head())
```

```python
print("\nРазмер таблицы:")
print(df.shape)
```

```python
print("\nТипы данных и количество непустых значений:")
df.info()
```

```python
print("\nКоличество транзакций каждого класса:")
print(df["Class"].value_counts())
```

```python
print("\nДоля каждого класса в процентах:")
print(df["Class"].value_counts(normalize=True) * 100)
```

```python
print("\nКоличество полных дубликатов:")
print(df.duplicated().sum())
```

```python
duplicates = df[df.duplicated()]
print("Количество повторов по классам:")
print(duplicates["Class"].value_counts())
```

```python
df_clean = df.drop_duplicates().copy()

print("Размер после удаления повторов:", df_clean.shape)
print("Осталось дубликатов:", df_clean.duplicated().sum())
print("Количество операций по классам:")
print(df_clean["Class"].value_counts())
```

```python
df_clean.to_csv("../data/creditcard_clean.csv", index=False)
```
